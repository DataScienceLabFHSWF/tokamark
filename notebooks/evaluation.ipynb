### Joining results from several models

```python
from pathlib import Path
import pandas as pd
```

```python
base_model = ('persistence', '/home/ir-zaya1/fusion/fairmast-data-preprocessing/output/persistence/')
models_to_join = {
    'cnn_23_01' : '/home/ir-zaya1/fusion/fairmast-data-preprocessing/output/cnn_23_01/'
}
```

```python
def join_results(metrics_file, metric_columns, index_column):
    def rename_cols(cols, suffix):
        return {col : col+'_'+suffix for col in cols}

    df_groups_joined  = pd.read_csv(
        Path(base_model[1])/metrics_file, index_col=index_column)\
        [metric_columns]\
        .rename(columns=rename_cols(metric_columns, base_model[0]))

    for model in models_to_join:
        df_groups = pd.read_csv(
            Path(models_to_join[model])/metrics_file, index_col=index_column)\
            [metric_columns]\
            .rename(columns=rename_cols(metric_columns, model))

        df_groups_joined = df_groups_joined.join(
            df_groups,
            how='outer'
        )
    return df_groups_joined

df_groups_joined = join_results(metrics_file='groups_metrics.csv', metric_columns=['NRMSE'], index_column='task')
df_groups_joined
```

```python
df_signals_joined = join_results(metrics_file='signals_metrics.csv', metric_columns=['RMSE','NRMSE'], index_column=['task', 'feature_name'])
df_signals_joined
```
