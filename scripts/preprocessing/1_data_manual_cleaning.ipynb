```python
import pandas as pd
import yaml

from preproc_paths import SHOT_SUMMARY_FILE
```

## Import expert cleaning 

```python
shot_summary = pd.read_csv(SHOT_SUMMARY_FILE)
shot_summary.columns
```

```python
map_flag_to_data = {
    'thomson_max_outlier': ['thomson_scattering-n_e', 'thomson_scattering-t_e'],
    'thomson_nan_count_outlier': ['thomson_scattering-n_e', 'thomson_scattering-t_e'],
    'sxr_max_outlier': ['soft_x_rays-horizontal_cam_lower', 'soft_x_rays-horizontal_cam_upper'],
    'magnetics_bfield_probe_mean_outlier': ['magnetics-b_field_pol_probe_ccbv_field',
                                            'magnetics-b_field_pol_probe_obr_field',
                                            'magnetics-b_field_pol_probe_obv_field',
                                            'magnetics-b_field_pol_probe_omv_voltage',
                                            'magnetics-b_field_tor_probe_cc_field',
                                            'magnetics-b_field_tor_probe_saddle_voltage'
                                            ]
}
```

```python
# Build the dict only for shots that have variables
shot_dict = {}
for _, row in shot_summary.iterrows():
    shot_id = row['shot_id']
    variables = []
    for flag, var_list in map_flag_to_data.items():
        if row.get(flag, False):
            variables.extend(var_list)
    if variables:  # Only save if there are variables
        shot_dict[shot_id] = variables

with open("dict_manual_outlier.yaml", "w") as f_:
    yaml.dump(shot_dict, f_, sort_keys=False)
```
